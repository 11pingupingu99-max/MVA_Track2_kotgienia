**v1.0.1 — corrected OddsPath implementation**

This release supersedes v1.0.0 for ClinGen/OddsPath arithmetic.
It generates no biological data and does not rank drugs.


```python
def oddspath(p1,p2):
    return (p2*(1-p1))/((1-p2)*p1)

def perfect_sep_brnich(n_p,n_b):
    # One misclassified variant is added to each readout set for perfect separation.
    p1=(n_p+1)/(n_p+n_b+2)
    p2_ps3=n_p/(n_p+1)
    p2_bs3=1/(n_b+1)
    return p1,p2_ps3,oddspath(p1,p2_ps3),p2_bs3,oddspath(p1,p2_bs3)
```

```python
p1,p2_ps3,ps3,p2_bs3,bs3=perfect_sep_brnich(5,6)
assert abs(p1-6/13)<1e-12
assert abs(ps3-5.833333333333333)<1e-12
assert abs(bs3-0.19444444444444445)<1e-12
print("Frozen 5 P/LP + 6 B/LB:")
print("P1 =",p1)
print("PS3 OddsPath =",ps3)
print("BS3 OddsPath =",bs3)
print("Maximum prospective strength: Moderate")
```

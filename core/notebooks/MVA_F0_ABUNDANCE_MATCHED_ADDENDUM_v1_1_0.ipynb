# MVA Track 2 · F0 mechanism-rule addendum v1.1.0
## Replace the binary 30% abundance threshold with an abundance-matched functional curve

This is a **prospective design correction before result-bearing F0 data exist**.

The previous frozen implementation used:
- `BUBR1 <30% WT` as abundance-limited;
- `BUBR1 >=30% WT` with dysfunction as function-limited.

That threshold is retained only as historical provenance. It is **not** biologically sufficient to call quantity "preserved".


## Revised mechanism classes

- **ABUNDANCE_DOMINANT:** reduced abundance/half-life and N1002K function approaches WT when mutant protein amount is matched to WT.
- **FUNCTION_DOMINANT:** at matched protein abundance, N1002K retains a reproducible segregation/SAC/alignment defect.
- **MIXED:** N1002K both reduces abundance/half-life and retains a residual functional defect after abundance matching.
- **INDETERMINATE:** confidence intervals or assay quality do not distinguish these states.

No single `%WT abundance` cutoff determines the class.


## Core experiment: abundance–function response surface

For both WT BUB1B and N1002K, create an expression series targeting approximately:

`10%, 30%, 50%, 75%, 100% WT steady-state BUBR1 abundance`

Measure in each condition:

1. BUBR1 abundance and half-life;
2. chromosome missegregation per completed division (primary);
3. SAC response;
4. chromosome alignment;
5. completed divisions / mitotic output;
6. viability and selective loss.

The scientific question becomes:

> At the **same measured BUBR1 abundance**, does N1002K produce more segregation error than WT?

## Statistical model

Use completed divisions as the denominator. A binomial or beta-binomial model is preferred over comparing percentages without denominators.

A practical primary model:

`missegregation ~ spline(measured_abundance) + variant + variant × spline(measured_abundance)`

with biological replicate as a random/blocking effect where supported.

Interpretation:

- no variant term / interaction after abundance matching → supports abundance-dominant;
- persistent variant effect at matched abundance → supports function-dominant component;
- reduced abundance plus persistent matched-abundance defect → mixed mechanism.

WT-BUB1B restoration remains the **causal positive rescue**, but it is not sufficient by itself to prove that stabilizing N1002K will work.


## Drug admission rule after this addendum

A proteostasis drug is admitted only if:

1. N1002K abundance/half-life is reduced **and**
2. increasing N1002K abundance itself moves function toward WT on the abundance–function curve.

If abundance rises without functional improvement, a chaperone/stabilizer is not credited as causal rescue.

This addendum does not generate biological results and does not assign PS3/BS3.
